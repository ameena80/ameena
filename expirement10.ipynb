```python
#Count each word in a text
text=input("Enter the text:")
words=text.split()
print("text is:",text)
for w in set(words):
    print(w,":",words.count(w))
```

```python
#replace the value greater than 100 with "over"
n=list(map(int,input("Enter integers separated by spaces:").split()))
print("Numbers are:",n)
for i in range(len(n)):
    if n[i]>100:
        n[i]="over"
print("result is :",n)
```

```python
#Count the occurence of "a"
names=input("Enter first name:").split()
print("Names:",names)
count=0
for name in names:
    count +=name.lower().count('a')
print("Occurence of 'a':",count)
```

```python
#Find GCD of two Numbers
a=int(input("Enter first number:"))
b=int(input("Enter second number:"))
print("Enter the first number:",a)
print("Enter the second number:",b)
while b!=0:
    a,b=b,a%b
print("GCD=",a)
```
